```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['Delhi', 'Mumbai', 'Chennai']
})
```

```python
for index,row in df.iterrows():
    print(f"{row["Name"]} live in {row["City"]} & {row["Age"]} years old.")
```

```python
for row in df.itertuples():
    print(row)
    print(row.Index)
    print(f"{row.Name} live in {row.City} & {row.Age} years old.")
```

```python
for col in df.columns:
    print(col)
    print(df[col].values)
```

```python
df["Category"] = df["Age"].apply(lambda age : "Young" if age < 30 else "Old")
df
```

```python
df["Zone"] = df["City"].apply(lambda city: "Danger" if city=="Delhi" else "Healthy")
df
```

```python
def check(row):
    city = row["City"]
    Age = row["Age"]
    if city=="Delhi":
        if Age < 30:
            return True
        else: return False
    else: 
        return False
    

```

```python
df["Delhi and young"] = df.apply(check, axis=1)
```
