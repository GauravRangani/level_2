```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_1 = pd.read_excel("D:/Learning Data science AI ML/Practice/level_2/01-03-2026 data visualisation/online_retail_II.xlsx",sheet_name="Year 2009-2010")
```

```python
df_2 = pd.read_excel(r"..\01-03-2026 data visualisation\online_retail_II.xlsx",sheet_name="Year 2010-2011")
```

```python
df_1.head()
```

```python
df_2.head()
```

```python
df_1.info()
```

```python
df = pd.concat((df_1,df_2),ignore_index=True)
```

```python
df.info()
```

```python
df = df.dropna(subset=["Customer ID"])
```

```python
df.info()
```

```python
df = df[~df["Invoice"].astype(str).str.startswith("C")]
```

```python
df.info()
```

```python
df[500:800]
```

```python
df = df[df["Quantity"]>0]
```

```python
df.info()
```

```python
df = df[df["Price"]>0]
```

```python
df.info()
```

```python
df.head()
```

```python
df["Revenue"] = df["Quantity"]*df["Price"]
```

```python
df.head()
```

```python
df["Month"] = df["InvoiceDate"].dt.to_period("M").astype(str)
```

```python
df.head()
```

```python
monthly_revenue = df.groupby("Month")["Revenue"].sum()
```

```python
monthly_revenue.index
```

```python
plt.figure(figsize=(12,6))
plt.plot(monthly_revenue.index,monthly_revenue.values,marker="o")
plt.title("Monthly_revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.yticks(rotation=45)
plt.tight_layout()
plt.grid(True)
plt.show()

```

```python
plt.figure(figsize=(12,7))
plt.style.use("classic")
plt.plot(monthly_revenue.index,monthly_revenue.values,marker='o')

plt.title("Monthly revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.yticks(rotation=45)

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,7))
plt.style.use("classic")
plt.plot(monthly_revenue.index,monthly_revenue.values,marker='o')

for i, value in enumerate(monthly_revenue.values):
    plt.text(monthly_revenue.index[i],value+10000,f'{int(value)}',ha="center",fontsize=8)

plt.title("Monthly revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.yticks(rotation=45)

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
font = {
    'family' : 'serif',
    'color' : 'red',
    'weight' : 'bold',
    'size' : 10
}

plt.figure(figsize=(12,7))
plt.style.use("classic")
plt.plot(monthly_revenue.index,monthly_revenue.values,marker='o',color="green")

for i, value in enumerate(monthly_revenue.values):
    plt.text(monthly_revenue.index[i],value+10000,f'{int(value)}',ha="center",fontdict=font)

plt.title("Monthly revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.yticks(rotation=45)

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
font = {
    'family' : 'serif',
    'color' : 'red',
    'weight' : 'bold',
    'size' : 10
}

plt.figure(figsize=(12,8),facecolor="white")
plt.style.use("classic")
plt.plot(monthly_revenue.index,monthly_revenue.values,marker='o',color="green")

for i, value in enumerate(monthly_revenue.values):
    plt.text(monthly_revenue.index[i],value+10000,f'{int(value)}',ha="center",fontdict=font)

plt.plot(monthly_revenue.index,monthly_revenue.values+1000000,marker='o',color="blue")

for i, value in enumerate(monthly_revenue.values):
    plt.text(monthly_revenue.index[i],value+1100000,f'{int(value+1000000)}',ha="center",fontdict=font)

plt.title("Monthly revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.yticks(rotation=45)

plt.grid(True)
plt.tight_layout()
plt.show()
```
